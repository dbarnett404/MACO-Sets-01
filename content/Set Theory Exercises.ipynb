# Visualizing Sets with Math and Code

Mathematical symbols can look like a secret language, but they match Python operators perfectly. Use this quick reference:

| Mathematical Concept | Math Symbol | Python Code Equivalent | Meaning |
| :--- | :---: | :---: | :--- |
| **Intersection** | ∩ | `and` / `&` | Must be in **both** groups |
| **Union** | ∪ | `or` / &#124; | In **at least one** group |
| **Difference** | − | `-` | In the first group, **not** the second |
| **Element of** | ∈ | `in` | This value belongs to the set |

## Exercise A: Filtering the Continuous Universe

In Mathematics, we use **Set-Builder Notation** like {x ∈ ℝ : x ≥ 5} to define a set of real numbers.

Let's test if a number x belongs to this set or its **Complement** (Aᶜ).

**YOUR TASK:** Change the value of `x` below to a negative number, a decimal, or a giant number. Press **Shift + Enter** to test it!


```python
# 1. Change this number to test your set!
x = 3.5 

# 2. Check the condition: Is x greater than or equal to 5?
in_set_A = (x >= 5)
in_complement_A = not in_set_A

print(f"Is x={x} an element of {{x ∈ ℝ : x ≥ 5}}? ", in_set_A)
print(f"Is x={x} in the Complement (Aᶜ)?          ", in_complement_A)
```
