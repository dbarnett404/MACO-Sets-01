## 1. Sets have no order

Two lists with the same items in a different order are *not* equal, but two sets with the same elements *are* equal.

```python
# Two lists with the same items in a different order are NOT equal...
list1 = [1, 2, 3]
list2 = [3, 2, 1]
print("Lists equal?", list1 == list2)

# ...but two sets with the same elements ARE equal
set1 = {1, 2, 3}
set2 = {3, 2, 1}
print("Sets equal?", set1 == set2)
print("Set 1:", set1)
print("Set 2:", set2)

# TRY IT: change the numbers in set2 so the sets are no longer equal
```

## 2. Sets have no duplicates

Listing an element more than once doesn't change the set. Each element is either in the set or it isn't.

```python
languages = ["Python", "Java", "JavaScript", "Python", "Java", "Python"]
print("List:", languages)

unique_languages = set(languages)
print("Set: ", unique_languages)
print("Items in list:", len(languages))
print("Items in set: ", len(unique_languages))

# TRY IT: add another repeated language to the list and run again
```

## 3. Membership (∈)

Is something an element of a set?

```python
A = {1, 4, 9, 3}

print("1 ∈ A?", 1 in A)
print("10 ∈ A?", 10 in A)

# TRY IT: test a different number, e.g. print("4 ∈ A?", 4 in A)
```

## 4. Union (∪)

Everything that is in A **or** B (or both).

```python
A = {1, 2, 3}
B = {3, 4, 5}

print("A =", A)
print("B =", B)
print("A ∪ B =", A.union(B))

# TRY IT: change the numbers in A and B
```

## 5. Intersection (∩)

Only the elements that A and B have **in common**.

```python
A = {1, 2, 3}
B = {3, 4, 5}

print("A =", A)
print("B =", B)
print("A ∩ B =", A.intersection(B))

# TRY IT: change B so that A and B have nothing in common. What is printed?
```

## 6. Difference (−)

What is in A but **not** in B.

```python
A = {1, 2, 3}
B = {3, 4, 5}

print("A =", A)
print("B =", B)
print("A − B =", A.difference(B))
print("B − A =", B.difference(A))

# TRY IT: is A − B always the same as B − A?
```

## 7. Subsets (⊆) and the empty set (∅)

A is a **subset** of B if every element of A is also in B.

The **empty set** has no elements at all. Python shows it as `set()`. You may have seen this already in section 5, when two sets had nothing in common.

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

print("A ⊆ B?", A.issubset(B))
print("B ⊆ A?", B.issubset(A))

empty = set()
print("The empty set ∅ is shown as:", empty)
print("Number of elements in ∅:", len(empty))
print("∅ ⊆ A?", empty.issubset(A))

# TRY IT: add the number 9 to A. Is A still a subset of B? Why not?
```
