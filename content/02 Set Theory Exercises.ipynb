## Exercise A: Counting people, not sign-ups

A club has a sign-up sheet, and some people signed up more than once.

**Question: how many different people signed up?**

```python
sign_ups = ["Amir", "Beth", "Chen", "Beth", "Dana", "Amir", "Beth", "Eli"]
people = set(sign_ups)

print("Sign-ups received:", len(sign_ups))
print("Different people: ", len(people))
print("Who signed up:    ", sorted(people))

# TRY IT: add your own name to the list twice and run again.
# THINK: which number changed, and which didn't? Why?
```

## Exercise B: Two clubs

Some students are in the chess club, some in the coding club, and some in both.

**Questions:**
- Who is in both clubs?
- Who is in at least one club (for sending a newsletter)?
- Who is in chess but not coding?

```python
chess = {"Amir", "Beth", "Chen", "Dana"}
coding = {"Chen", "Dana", "Eli", "Farah"}

print("Chess club: ", sorted(chess))
print("Coding club:", sorted(coding))
print()
print("In both clubs (∩):         ", sorted(chess.intersection(coding)))
print("In at least one club (∪):  ", sorted(chess.union(coding)))
print("Chess only (chess − coding):", sorted(chess.difference(coding)))
print("Coding only (coding − chess):", sorted(coding.difference(chess)))

# TRY IT: add "Beth" to the coding club and run again. What changes?
# THINK: which of the four answers would you need to find the number of people to email?
```

## Exercise C: Can I take this course?

A course has **required** skills. A student can enrol only if **every** required skill is one they already have.

That is exactly the question "is the required set a **subset** of the student's skills?"

```python
required = {"algebra", "functions", "logic"}
student_has = {"algebra", "logic"}

print("Required:   ", sorted(required))
print("Student has:", sorted(student_has))
print()
print("Can enrol (required ⊆ student has)?", required.issubset(student_has))
print("Still missing (required − student has):", sorted(required.difference(student_has)))

# TRY IT: add "functions" to student_has and run again.
# TRY IT: now add "statistics" to student_has. Can they still enrol? Does having extra skills matter?
```

## Exercise D: Finding new music

Alex and Sam each have a set of songs they like.

**Questions:**
- Which songs do they both like?
- Which of Sam's songs could Alex discover?
- How many different songs are needed for a shared party playlist?

```python
alex = {"Sunrise", "Night Drive", "Echoes", "Gravity"}
sam = {"Echoes", "Gravity", "Daydream", "Neon", "Horizon"}

print("Both like (∩):             ", sorted(alex.intersection(sam)))
print("Alex could discover (sam − alex):", sorted(sam.difference(alex)))
print("Party playlist (∪):        ", sorted(alex.union(sam)))
print("Songs in the playlist:     ", len(alex.union(sam)))

# TRY IT: add two of your own made-up songs to Alex's set and one to Sam's. Run again.
# THINK: if a song is on both lists, does it appear once or twice in the party playlist? Why is that useful?
```

## Exercise E: Predict, then check

First, on paper, work out the answers for these sets:

1. A ∩ B
2. A ∪ C
3. (A ∪ B) ∩ C
4. A − (B ∪ C)

Then run the cell to check your answers.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
C = {4, 6, 8}

print("1. A ∩ B =", A.intersection(B))
print("2. A ∪ C =", A.union(C))
print("3. (A ∪ B) ∩ C =", A.union(B).intersection(C))
print("4. A − (B ∪ C) =", A.difference(B.union(C)))

# TRY IT: change one number in C. Before running, predict which answers will change.
```
